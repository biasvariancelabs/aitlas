### Define config_mlc_1 and run forward passes

```python
# Main loop
print(f"{'BACKBONE VARIATION':<60} | {'STATUS':<10} | {'MESSAGE'}")
print("-" * 150)

for bb_conf in backbone_configs:
    # Prepare the onfig
    config_mlc_1 = {
        **task_config,
        **bb_conf,
        #**neck_config,
        #**decoder_config,
        **head_config
    }
    
    name = config_mlc_1['backbone_name']
    
    # Define backbone variations
    variations = []
    
    if name == 'anysat_base':
        dates = torch.zeros(1).unsqueeze(1).to(device)
        patch_size = 160
        data = {
            "s2": dummy_input_6,
            "s2_dates": dates
        }
        variations.append((f"{name} (output=tile)", {"x": data, "patch_size": patch_size, "output": "tile"}))
        variations.append((f"{name} (output=patch)", {"x": data, "patch_size": patch_size, "output": "patch"}))
        variations.append((f"{name} (output=all)", {"x": data, "patch_size": patch_size, "output": "all"}))
        variations.append((f"{name} (output=dense)", {"x": data, "patch_size": patch_size, "output": "dense"}))

    elif name in ['copernicusfm_base', 'copernicusfm_large']:
        metadata = torch.randn(1, 4).to(device)
        wavelengths_copernicus = [x * 1000 for x in wavelengths]
        bandwidths_copernicus = [x * 1000 for x in bandwidths]
        variations.append((name, {"x": dummy_input_4, "metadata": metadata, "wavelengths": wavelengths_copernicus, "bandwidths": bandwidths_copernicus}))

    elif name in ['croma_base', 'croma_large']:
        variations.append((f"{name} (optical)", {"x_optical": dummy_input_8, "modalities": ['optical']}))
        variations.append((f"{name} (sar)", {"x_sar": dummy_input_9, "modalities": ['sar']}))
        variations.append((f"{name} (both)", {"x_optical": dummy_input_8, "x_sar": dummy_input_9, "modalities": ['optical', 'sar']}))

    elif name in ['dofa_v1_vit_base_patch16', 'dofa_v1_vit_large_patch16', 'dofa_v2_vit_base_patch14', 'dofa_v2_vit_large_patch14']:
        variations.append((name, {"x": dummy_input_4, "wave_list": wavelengths}))
        
    elif name in ['galileo_nano', 'galileo_tiny', 'galileo_base']:
        data = {
            "s2": dummy_input_6
        }
        variations.append((f"{name} (average=True)", {"inputs": data, "average_features": True}))
        variations.append((f"{name} (average=False)", {"inputs": data, "average_features": False}))
    
    elif name in ['gassl_moco_resnet50', 'gassl_moco_geo_resnet50']:
        variations.append((f"{name} (all=True)",  {"x": dummy_input_1, "return_all_embeddings": True}))
        variations.append((f"{name} (all=False)", {"x": dummy_input_1, "return_all_embeddings": False}))
    
    elif name == 'panopticon_base':
        wavelengths_panopticon = [int(1000*bands[key]) for key in bands.keys()] 
        wavelengths_panopticon_tensor = torch.tensor([wavelengths_panopticon] * dummy_input_3.shape[0]) 
        wavelengths_panopticon_tensor = torch.cat((wavelengths_panopticon_tensor[:, :9], wavelengths_panopticon_tensor[:, 10:]), dim=1).to(device)
        data = {
            "imgs": dummy_input_3,
            "chn_ids": wavelengths_panopticon_tensor
        }
        variations.append((f"{name} (dense=True)", {"x_dict": data, "dense_features": True}))
        variations.append((f"{name} (dense=False)", {"x_dict": data, "dense_features": False}))

    elif name == 'presto_default':
        latlons = torch.randn(1, 2, 224, 224).to(device)
        s2_presto = dummy_input_6.to(device)
        data = {
            "s2": s2_presto,
            "latlons": latlons
        }
        variations.append((f"{name}", {"inputs": data}))

    elif name in ['prithvi_eo_v2_base_tl', 'prithvi_eo_v2_large', 'prithvi_eo_v2_large_tl', 'prithvi_eo_v2_huge', 'prithvi_eo_v2_huge_tl']:
        variations.append((f"{name} (seg_features=True)", {"x": dummy_input_10, "segmentation_features": True}))
        variations.append((f"{name} (seg_features=False)", {"x": dummy_input_10, "segmentation_features": False}))

    elif name == 'prithvi_eo_v1_base':
        variations.append((f"{name} (seg_features=True)", {"x": dummy_input_11, "segmentation_features": True}))
        variations.append((f"{name} (seg_features=False)", {"x": dummy_input_11, "segmentation_features": False}))

    elif name == 'prithvi_eo_v2_tiny_tl':
        variations.append((f"{name} (seg_features=True)", {"x": dummy_input_12, "segmentation_features": True}))
        variations.append((f"{name} (seg_features=False)", {"x": dummy_input_12, "segmentation_features": False}))
    
    elif name in ['satmae_vit_base_multispectral', 'satmae_vit_large_multispectral']:
        variations.append((name, {"x": dummy_input_2}))

    elif name == 'satmae_vit_large_temporal':
        variations.append((f"{name} (Skipped)", None))
        
    elif name == 'satmae_plusplus_vit_large_multispectral':
        variations.append((name, {"x": dummy_input_5}))
        
    elif name == 'scalemae_vit_large_patch16':
        input_res = torch.tensor([1], device=device)
        variations.append((name, {"x": dummy_input_1, "input_res": input_res}))

    elif name == 'terrafm_base':
        variations.append((name, {"x": dummy_input_3}))

    elif name in ['terramind_v1_tiny', 'terramind_v1_small', 'terramind_v1_base', 'terramind_v1_large']:
        data = {
            "S2L2A": dummy_input_3
        }
        variations.append((f"{name} (merge_method=mean)", {"x": data, "merge_method": "mean"}))
        variations.append((f"{name} (merge_method=max)", {"x": data, "merge_method": "max"}))
        variations.append((f"{name} (merge_method=concat)", {"x": data, "merge_method": "concat"}))
        variations.append((f"{name} (merge_method=dict)", {"x": data, "merge_method": "dict"}))

    elif name in ['terramind_v1_tiny_tim', 'terramind_v1_small_tim', 'terramind_v1_base_tim', 'terramind_v1_large_tim']:
        data = {
            "S2L2A": dummy_input_3
        }
        variations.append((f"{name} (merge_method=mean)", {"x": data, "merge_method": "mean"}))
        variations.append((f"{name} (merge_method=max)", {"x": data, "merge_method": "max"}))
        variations.append((f"{name} (merge_method=concat)", {"x": data, "merge_method": "concat"}))
        variations.append((f"{name} (merge_method=dict)", {"x": data, "merge_method": "dict"}))
        
    else:
        # Default RGB case
        variations.append((name, {"x": dummy_input_1}))

    # Execution loop
    model_instantiation_failed = False
    try:
        with suppress_output():
            model = CompositeModel(config_mlc_1)
            model.to(device)
            model.eval()
    except Exception as e:
        model_instantiation_failed = True
        print(f"{name:<60} | {'ERROR':<10} | Init Failed: {str(e)}")

    if not model_instantiation_failed:
        for label, kwargs in variations:
            if kwargs is None: # Handle skipped cases
                 print(f"{label:<60} | {'SKIP':<10} | Skipped by config")
                 continue

            try:
                with suppress_output():
                    with torch.no_grad():
                        features = model.forward(**kwargs) # Use model.predict for prediction tasks (output predictions/values instead of logits)
                
                # If we get here, it succeeded
                print(f"{label:<60} | {'OK':<10} | Success")

            except Exception as e:
                # Capture the specific error for this variation
                print(f"{label:<60} | {'ERROR':<10} | {str(e)}")

print("-" * 150)
```
