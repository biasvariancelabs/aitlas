## Import the required packages

```python
import warnings


warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=DeprecationWarning)

from aitlas.datasets import ObjectDetectionRotatedParallaxDataset
```

## Visualize images and objects with bounding boxes

```python
dataset_config = {
    "data_dir": "/home/dragik/data/parallax/TCI_jpg/",
    "annotation_dir": "/home/dragik/data/parallax/Parallax_COCO/",
}
dataset = ObjectDetectionRotatedParallaxDataset(dataset_config)

print(f"Total number of patches: {len(dataset)}")
dataset.show_image(250)
dataset.show_image(50)
dataset.show_image(100);
```

```python
dataset.show_batch(15);
```

```python
dataset.data_distribution_table()
```
